```python
import pandas as pd
import numpy as np
import time
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# MODEL COMPARISON CLASS - TRACKS AND DISPLAYS ALL METRICS
# ============================================================================

class ModelComparison:
    """
    Comprehensive model comparison tool for flight delay prediction models.
    Tracks MAE, MSE, RMSE, R², MAPE, SMAPE, prediction speed, and training time.
    """
    
    def __init__(self):
        self.results = []
        self.models = {}
        
    def calculate_mape(self, y_true, y_pred):
        """Mean Absolute Percentage Error"""
        y_true, y_pred = np.array(y_true), np.array(y_pred)
        mask = y_true != 0
        if mask.sum() == 0:
            return 0.0
        return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100
    
    def calculate_smape(self, y_true, y_pred):
        """Symmetric Mean Absolute Percentage Error"""
        y_true, y_pred = np.array(y_true), np.array(y_pred)
        denominator = (np.abs(y_true) + np.abs(y_pred)) / 2
        mask = denominator != 0
        if mask.sum() == 0:
            return 0.0
        return np.mean(np.abs(y_true[mask] - y_pred[mask]) / denominator[mask]) * 100
    
    def evaluate_model(self, model, X_test, y_dep_test, y_arr_test, model_name, training_time):
        """
        Comprehensive evaluation of a model on test data.
        """
        print(f"\n📊 Evaluating {model_name}...")
        
        # Measure prediction time
        start_time = time.time()
        
        # Make predictions
        predictions = model.predict(X_test, verbose=0)
        
        # Handle different prediction formats
        if isinstance(predictions, list):
            y_dep_pred = predictions[0].flatten()
            y_arr_pred = predictions[1].flatten()
        else:
            y_dep_pred = predictions.flatten()
            y_arr_pred = predictions.flatten()
        
        prediction_time = time.time() - start_time
        predictions_per_second = len(X_test) / prediction_time
        
        # Calculate metrics for departure delay
        dep_metrics = self.calculate_metrics(y_dep_test, y_dep_pred)
        
        # Calculate metrics for arrival delay
        arr_metrics = self.calculate_metrics(y_arr_test, y_arr_pred)
        
        # Store results
        result = {
            'model_name': model_name,
            'training_time_seconds': training_time,
            'training_time_minutes': training_time / 60,
            'prediction_time_seconds': prediction_time,
            'predictions_per_second': predictions_per_second,
            'samples_test': len(X_test),
            
            # Departure metrics
            'dep_mae': dep_metrics['mae'],
            'dep_mse': dep_metrics['mse'],
            'dep_rmse': dep_metrics['rmse'],
            'dep_r2': dep_metrics['r2'],
            'dep_mape': dep_metrics['mape'],
            'dep_smape': dep_metrics['smape'],
            
            # Arrival metrics
            'arr_mae': arr_metrics['mae'],
            'arr_mse': arr_metrics['mse'],
            'arr_rmse': arr_metrics['rmse'],
            'arr_r2': arr_metrics['r2'],
            'arr_mape': arr_metrics['mape'],
            'arr_smape': arr_metrics['smape'],
            
            # Combined metrics (average)
            'avg_mae': (dep_metrics['mae'] + arr_metrics['mae']) / 2,
            'avg_rmse': (dep_metrics['rmse'] + arr_metrics['rmse']) / 2,
            'avg_r2': (dep_metrics['r2'] + arr_metrics['r2']) / 2,
            'avg_mape': (dep_metrics['mape'] + arr_metrics['mape']) / 2,
            'avg_smape': (dep_metrics['smape'] + arr_metrics['smape']) / 2,
        }
        
        self.results.append(result)
        self.models[model_name] = model
        
        # Display immediate results for this model
        self.display_model_summary(result)
        
        return result
    
    def calculate_metrics(self, y_true, y_pred):
        """Calculate all metrics for a single target"""
        return {
            'mae': mean_absolute_error(y_true, y_pred),
            'mse': mean_squared_error(y_true, y_pred),
            'rmse': np.sqrt(mean_squared_error(y_true, y_pred)),
            'r2': r2_score(y_true, y_pred),
            'mape': self.calculate_mape(y_true, y_pred),
            'smape': self.calculate_smape(y_true, y_pred)
        }
    
    def display_model_summary(self, result):
        """Display summary for a single model"""
        print("\n" + "─" * 90)
        print(f"📈 MODEL PERFORMANCE SUMMARY: {result['model_name']}")
        print("─" * 90)
        print(f"⏱️  Training Time: {result['training_time_minutes']:.2f} minutes")
        print(f"⚡ Prediction Speed: {result['predictions_per_second']:,.0f} predictions/second")
        print(f"📊 Test Samples: {result['samples_test']:,}")
        print()
        print(f"   {'Metric':<20} {'Departure':>15} {'Arrival':>15} {'Average':>15}")
        print("   " + "-" * 68)
        print(f"   {'MAE (min)':<20} {result['dep_mae']:>15.2f} {result['arr_mae']:>15.2f} {result['avg_mae']:>15.2f}")
        print(f"   {'RMSE (min)':<20} {result['dep_rmse']:>15.2f} {result['arr_rmse']:>15.2f} {result['avg_rmse']:>15.2f}")
        print(f"   {'R² Score':<20} {result['dep_r2']:>15.4f} {result['arr_r2']:>15.4f} {result['avg_r2']:>15.4f}")
        print(f"   {'MAPE (%)':<20} {result['dep_mape']:>15.2f} {result['arr_mape']:>15.2f} {result['avg_mape']:>15.2f}")
        print(f"   {'SMAPE (%)':<20} {result['dep_smape']:>15.2f} {result['arr_smape']:>15.2f} {result['avg_smape']:>15.2f}")
        print("─" * 90)
    
    def display_comparison_table(self):
        """
        Display a formatted comparison table of all models including SMAPE.
        """
        if not self.results:
            print("No results to display. Run evaluate_model() first.")
            return
        
        # Convert to DataFrame
        df = pd.DataFrame(self.results)
        
        print("\n" + "=" * 140)
        print("🚀 FLIGHT DELAY PREDICTION - COMPLETE MODEL COMPARISON TABLE (WITH SMAPE)")
        print("=" * 140)
        print()
        
        # Format the main comparison table
        pd.set_option('display.max_columns', None)
        pd.set_option('display.width', 160)
        pd.set_option('display.float_format', '{:.3f}'.format)
        
        # Create a formatted display DataFrame with SMAPE included
        display_data = []
        for _, row in df.iterrows():
            display_data.append({
                'Model': row['model_name'],
                'Train (min)': f"{row['training_time_minutes']:.1f}",
                'Pred/Sec': f"{row['predictions_per_second']:,.0f}",
                'Dep MAE': f"{row['dep_mae']:.2f}",
                'Dep RMSE': f"{row['dep_rmse']:.2f}",
                'Dep R²': f"{row['dep_r2']:.3f}",
                'Dep MAPE%': f"{row['dep_mape']:.1f}",
                'Dep SMAPE%': f"{row['dep_smape']:.1f}",
                'Arr MAE': f"{row['arr_mae']:.2f}",
                'Arr RMSE': f"{row['arr_rmse']:.2f}",
                'Arr R²': f"{row['arr_r2']:.3f}",
                'Arr MAPE%': f"{row['arr_mape']:.1f}",
                'Arr SMAPE%': f"{row['arr_smape']:.1f}",
                'Avg MAE': f"{row['avg_mae']:.2f}",
                'Avg RMSE': f"{row['avg_rmse']:.2f}",
                'Avg R²': f"{row['avg_r2']:.3f}",
                'Avg MAPE%': f"{row['avg_mape']:.1f}",
                'Avg SMAPE%': f"{row['avg_smape']:.1f}"
            })
        
        display_df = pd.DataFrame(display_data)
        print(display_df.to_string(index=False))
        
        # Find best model for each metric
        self.display_best_models(df)
        
        return df
    
    def display_best_models(self, df):
        """Display which model performs best for each metric including SMAPE"""
        print("\n" + "=" * 100)
        print("🏆 BEST PERFORMING MODELS BY METRIC (WITH SMAPE)")
        print("=" * 100)
        
        metrics_to_compare = {
            'Departure MAE': ('dep_mae', 'lower'),
            'Departure RMSE': ('dep_rmse', 'lower'),
            'Departure R²': ('dep_r2', 'higher'),
            'Departure MAPE': ('dep_mape', 'lower'),
            'Departure SMAPE': ('dep_smape', 'lower'),
            'Arrival MAE': ('arr_mae', 'lower'),
            'Arrival RMSE': ('arr_rmse', 'lower'),
            'Arrival R²': ('arr_r2', 'higher'),
            'Arrival MAPE': ('arr_mape', 'lower'),
            'Arrival SMAPE': ('arr_smape', 'lower'),
            'Average MAE': ('avg_mae', 'lower'),
            'Average RMSE': ('avg_rmse', 'lower'),
            'Average R²': ('avg_r2', 'higher'),
            'Average MAPE': ('avg_mape', 'lower'),
            'Average SMAPE': ('avg_smape', 'lower'),
            'Training Time': ('training_time_minutes', 'lower'),
            'Prediction Speed': ('predictions_per_second', 'higher')
        }
        
        # Create a formatted table for best models
        best_models_data = []
        for metric_name, (col, goal) in metrics_to_compare.items():
            if goal == 'lower':
                best_idx = df[col].idxmin()
                best_value = df.loc[best_idx, col]
                best_model = df.loc[best_idx, 'model_name']
            else:
                best_idx = df[col].idxmax()
                best_value = df.loc[best_idx, col]
                best_model = df.loc[best_idx, 'model_name']
            
            # Format value for display
            if 'R²' in metric_name:
                formatted_value = f"{best_value:.4f}"
            elif 'MAPE' in metric_name or 'SMAPE' in metric_name:
                formatted_value = f"{best_value:.2f}%"
            elif 'Training' in metric_name:
                formatted_value = f"{best_value:.2f} min"
            elif 'Prediction' in metric_name:
                formatted_value = f"{best_value:,.0f}/sec"
            else:
                formatted_value = f"{best_value:.2f}"
            
            best_models_data.append({
                'Metric': metric_name,
                'Best Model': best_model,
                'Value': formatted_value
            })
        
        best_df = pd.DataFrame(best_models_data)
        print(best_df.to_string(index=False))
    
    def plot_comparison(self):
        """
        Create visualization comparing model performance with MAPE vs SMAPE comparison.
        """
        try:
            import matplotlib.pyplot as plt
            import seaborn as sns
            
            df = pd.DataFrame(self.results)
            
            # Create figure with 2x5 subplots (10 total plots)
            fig, axes = plt.subplots(2, 5, figsize=(24, 12))
            fig.suptitle('Flight Delay Prediction Models - Comprehensive Performance Comparison\n(With MAPE vs SMAPE)', 
                        fontsize=18, fontweight='bold')
            
            x = np.arange(len(df))
            width = 0.35
            
            # 1. MAE Comparison
            ax = axes[0, 0]
            ax.bar(x - width/2, df['dep_mae'], width, label='Departure', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_mae'], width, label='Arrival', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('MAE (minutes)')
            ax.set_title('Mean Absolute Error')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 2. RMSE Comparison
            ax = axes[0, 1]
            ax.bar(x - width/2, df['dep_rmse'], width, label='Departure', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_rmse'], width, label='Arrival', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('RMSE (minutes)')
            ax.set_title('Root Mean Square Error')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 3. R² Comparison
            ax = axes[0, 2]
            ax.bar(x - width/2, df['dep_r2'], width, label='Departure', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_r2'], width, label='Arrival', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('R² Score')
            ax.set_title('R² Score')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 4. MAPE Comparison
            ax = axes[0, 3]
            ax.bar(x - width/2, df['dep_mape'], width, label='Departure', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_mape'], width, label='Arrival', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('MAPE (%)')
            ax.set_title('Mean Absolute Percentage Error')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 5. SMAPE Comparison
            ax = axes[0, 4]
            ax.bar(x - width/2, df['dep_smape'], width, label='Departure', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_smape'], width, label='Arrival', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('SMAPE (%)')
            ax.set_title('Symmetric Mean Absolute Percentage Error')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 6. Training Time
            ax = axes[1, 0]
            colors = ['gold' if v == df['training_time_minutes'].min() else 'lightblue' 
                     for v in df['training_time_minutes']]
            ax.bar(df['model_name'], df['training_time_minutes'], color=colors, alpha=0.8)
            ax.set_xlabel('Model')
            ax.set_ylabel('Training Time (minutes)')
            ax.set_title('Training Time Comparison')
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.grid(True, alpha=0.3)
            
            # 7. Prediction Speed
            ax = axes[1, 1]
            colors = ['gold' if v == df['predictions_per_second'].max() else 'lightgreen' 
                     for v in df['predictions_per_second']]
            ax.bar(df['model_name'], df['predictions_per_second'], color=colors, alpha=0.8)
            ax.set_xlabel('Model')
            ax.set_ylabel('Predictions/Second')
            ax.set_title('Inference Speed')
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.grid(True, alpha=0.3)
            
            # 8. MAPE vs SMAPE - Departure
            ax = axes[1, 2]
            ax.plot(df['model_name'], df['dep_mape'], 'o-', label='MAPE', linewidth=2, markersize=8, color='blue')
            ax.plot(df['model_name'], df['dep_smape'], 's-', label='SMAPE', linewidth=2, markersize=8, color='red')
            ax.set_xlabel('Model')
            ax.set_ylabel('Error (%)')
            ax.set_title('Departure: MAPE vs SMAPE')
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 9. MAPE vs SMAPE - Arrival
            ax = axes[1, 3]
            ax.plot(df['model_name'], df['arr_mape'], 'o-', label='MAPE', linewidth=2, markersize=8, color='blue')
            ax.plot(df['model_name'], df['arr_smape'], 's-', label='SMAPE', linewidth=2, markersize=8, color='red')
            ax.set_xlabel('Model')
            ax.set_ylabel('Error (%)')
            ax.set_title('Arrival: MAPE vs SMAPE')
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # 10. Average MAE vs RMSE
            ax = axes[1, 4]
            ax.plot(df['model_name'], df['avg_mae'], 'o-', label='Avg MAE', linewidth=2, markersize=8, color='green')
            ax.plot(df['model_name'], df['avg_rmse'], 's-', label='Avg RMSE', linewidth=2, markersize=8, color='orange')
            ax.set_xlabel('Model')
            ax.set_ylabel('Minutes')
            ax.set_title('Average MAE vs RMSE')
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right', fontsize=9)
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            plt.tight_layout()
            plt.show()
            
            # Create additional figure for detailed MAPE vs SMAPE comparison
            fig2, axes2 = plt.subplots(1, 2, figsize=(16, 6))
            fig2.suptitle('MAPE vs SMAPE - Detailed Comparison Across Models', fontsize=14, fontweight='bold')
            
            # Departure - MAPE vs SMAPE bar comparison
            ax = axes2[0]
            x = np.arange(len(df))
            width = 0.35
            ax.bar(x - width/2, df['dep_mape'], width, label='MAPE', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['dep_smape'], width, label='SMAPE', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('Percentage Error (%)')
            ax.set_title('Departure Delay - MAPE vs SMAPE')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right')
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            # Arrival - MAPE vs SMAPE bar comparison
            ax = axes2[1]
            ax.bar(x - width/2, df['arr_mape'], width, label='MAPE', alpha=0.8, color='steelblue')
            ax.bar(x + width/2, df['arr_smape'], width, label='SMAPE', alpha=0.8, color='coral')
            ax.set_xlabel('Model')
            ax.set_ylabel('Percentage Error (%)')
            ax.set_title('Arrival Delay - MAPE vs SMAPE')
            ax.set_xticks(x)
            ax.set_xticklabels(df['model_name'], rotation=45, ha='right')
            ax.legend()
            ax.grid(True, alpha=0.3)
            
            plt.tight_layout()
            plt.show()
            
        except ImportError as e:
            print(f"⚠️  Could not generate plots: {e}")
            print("   Install matplotlib and seaborn for visualizations:")
            print("   pip install matplotlib seaborn")
    
    def save_comparison_report(self, filename='model_comparison_report'):
        """
        Save comparison results to CSV and text files with SMAPE included.
        """
        df = pd.DataFrame(self.results)
        
        # Save CSV
        csv_filename = f'{filename}.csv'
        df.to_csv(csv_filename, index=False)
        print(f"\n💾 CSV report saved to '{csv_filename}'")
        
        # Save detailed text report with SMAPE
        txt_filename = f'{filename}_summary.txt'
        with open(txt_filename, 'w', encoding='utf-8') as f:
            f.write("=" * 100 + "\n")
            f.write("FLIGHT DELAY PREDICTION - COMPREHENSIVE MODEL COMPARISON REPORT (WITH SMAPE)\n")
            f.write("=" * 100 + "\n\n")
            f.write(f"Generated: {time.strftime('%Y-%m-%d %H:%M:%S')}\n")
            f.write(f"Total Models Compared: {len(self.results)}\n\n")
            
            for _, row in df.iterrows():
                f.write(f"\n{'─' * 100}\n")
                f.write(f"MODEL: {row['model_name']}\n")
                f.write(f"{'─' * 100}\n")
                f.write(f"Training Time: {row['training_time_minutes']:.2f} minutes\n")
                f.write(f"Prediction Speed: {row['predictions_per_second']:,.0f} predictions/second\n")
                f.write(f"Test Samples: {row['samples_test']:,}\n\n")
                
                f.write("METRICS SUMMARY (WITH SMAPE):\n")
                f.write(f"{'Metric':<20} {'Departure':>15} {'Arrival':>15} {'Average':>15}\n")
                f.write(f"{'-' * 68}\n")
                f.write(f"{'MAE (min)':<20} {row['dep_mae']:>15.2f} {row['arr_mae']:>15.2f} {row['avg_mae']:>15.2f}\n")
                f.write(f"{'RMSE (min)':<20} {row['dep_rmse']:>15.2f} {row['arr_rmse']:>15.2f} {row['avg_rmse']:>15.2f}\n")
                f.write(f"{'R² Score':<20} {row['dep_r2']:>15.4f} {row['arr_r2']:>15.4f} {row['avg_r2']:>15.4f}\n")
                f.write(f"{'MAPE (%)':<20} {row['dep_mape']:>15.2f} {row['arr_mape']:>15.2f} {row['avg_mape']:>15.2f}\n")
                f.write(f"{'SMAPE (%)':<20} {row['dep_smape']:>15.2f} {row['arr_smape']:>15.2f} {row['avg_smape']:>15.2f}\n")
            
            # Best models summary with SMAPE
            f.write("\n\n" + "=" * 100 + "\n")
            f.write("BEST PERFORMING MODELS SUMMARY (WITH SMAPE)\n")
            f.write("=" * 100 + "\n\n")
            
            metrics_to_compare = {
                'Departure MAE': ('dep_mae', 'lower'),
                'Departure RMSE': ('dep_rmse', 'lower'),
                'Departure R²': ('dep_r2', 'higher'),
                'Departure MAPE': ('dep_mape', 'lower'),
                'Departure SMAPE': ('dep_smape', 'lower'),
                'Arrival MAE': ('arr_mae', 'lower'),
                'Arrival RMSE': ('arr_rmse', 'lower'),
                'Arrival R²': ('arr_r2', 'higher'),
                'Arrival MAPE': ('arr_mape', 'lower'),
                'Arrival SMAPE': ('arr_smape', 'lower'),
                'Average MAE': ('avg_mae', 'lower'),
                'Average RMSE': ('avg_rmse', 'lower'),
                'Average R²': ('avg_r2', 'higher'),
                'Average MAPE': ('avg_mape', 'lower'),
                'Average SMAPE': ('avg_smape', 'lower'),
                'Training Time': ('training_time_minutes', 'lower'),
                'Prediction Speed': ('predictions_per_second', 'higher')
            }
            
            for metric_name, (col, goal) in metrics_to_compare.items():
                if goal == 'lower':
                    best_idx = df[col].idxmin()
                    best_value = df.loc[best_idx, col]
                    best_model = df.loc[best_idx, 'model_name']
                else:
                    best_idx = df[col].idxmax()
                    best_value = df.loc[best_idx, col]
                    best_model = df.loc[best_idx, 'model_name']
                
                if 'R²' in metric_name:
                    f.write(f"{metric_name:25}: {best_model:30} → {best_value:.4f}\n")
                elif 'MAPE' in metric_name or 'SMAPE' in metric_name:
                    f.write(f"{metric_name:25}: {best_model:30} → {best_value:.2f}%\n")
                elif 'Training' in metric_name:
                    f.write(f"{metric_name:25}: {best_model:30} → {best_value:.2f} min\n")
                elif 'Prediction' in metric_name:
                    f.write(f"{metric_name:25}: {best_model:30} → {best_value:,.0f}/sec\n")
                else:
                    f.write(f"{metric_name:25}: {best_model:30} → {best_value:.2f}\n")
        
        print(f"💾 Text report saved to '{txt_filename}'")


# ============================================================================
# FLIGHT DELAY NEURAL NETWORK WITH MULTIPLE ARCHITECTURES
# ============================================================================

class FlightDelayNeuralNetwork:
    """
    Neural Network model for predicting flight departure and arrival delays
    with three different architectures optimized for 6.7M records.
    """
    
    def __init__(self, model_type='wide_deep'):
        """
        Initialize with specified model architecture.
        
        Args:
            model_type: 'original', 'wide_deep', 'residual', or 'transformer'
        """
        self.model = None
        self.model_type = model_type
        self.scaler = StandardScaler()
        self.label_encoders = {}
        self.feature_columns = None
        self.categorical_columns = []
        self.numerical_columns = []
        self.dataset_columns_info = None
        self.history = None
    
    def load_and_prepare_data(self):
        """Load the dataset and prepare it for training."""
        print("=" * 70)
        print("LOADING DATASET")
        print("=" * 70)
        
        # Load the dataset
        df = pd.read_csv('./dataset/df_cleaned.csv')
        print(f"Dataset loaded: {df.shape[0]:,} rows, {df.shape[1]} columns")
        
        # Store dataset info
        self.dataset_columns_info = {
            'all_columns': df.columns.tolist(),
            'dtypes': df.dtypes.to_dict()
        }
        
        print("\nDataset columns:")
        for i, col in enumerate(df.columns.tolist()[:20]):
            print(f"{i:2d}. {col}")
        if len(df.columns) > 20:
            print(f"  ... and {len(df.columns) - 20} more columns")
        
        return df
    
    def prepare_features(self, df):
        """Prepare features for neural network training."""
        df_processed = df.copy()
        
        print("\n" + "=" * 70)
        print("FEATURE PREPARATION")
        print("=" * 70)
        
        # Store target variables
        if 'dep_delay_min' in df_processed.columns:
            y_departure = df_processed['dep_delay_min'].values
            df_processed = df_processed.drop('dep_delay_min', axis=1)
            print(f"Target: dep_delay_min")
        else:
            raise ValueError("'dep_delay_min' column not found")
            
        if 'arr_delay_min' in df_processed.columns:
            y_arrival = df_processed['arr_delay_min'].values
            df_processed = df_processed.drop('arr_delay_min', axis=1)
            print(f"Target: arr_delay_min")
        else:
            raise ValueError("'arr_delay_min' column not found")
        
        # Remove other delay-related columns
        delay_columns = ['dep_delay_tag', 'dep_delay_type', 'arr_delay_type']
        for col in delay_columns:
            if col in df_processed.columns:
                df_processed = df_processed.drop(col, axis=1)
                print(f"Dropped: {col}")
        
        # Drop latitude and longitude columns
        geo_columns_to_drop = [
            'dep_latitude', 'dep_longitude', 
            'arr_latitude', 'arr_longitude'
        ]
        
        for col in geo_columns_to_drop:
            if col in df_processed.columns:
                df_processed = df_processed.drop(col, axis=1)
                print(f"Dropped: {col}")
        
        # Handle date column
        if 'flight_date' in df_processed.columns:
            print("\nProcessing flight_date column...")
            df_processed['flight_date'] = pd.to_datetime(df_processed['flight_date'])
            df_processed['flight_year'] = df_processed['flight_date'].dt.year
            df_processed['flight_month'] = df_processed['flight_date'].dt.month
            df_processed['flight_day'] = df_processed['flight_date'].dt.day
            df_processed['flight_dayofweek'] = df_processed['flight_date'].dt.dayofweek
            df_processed['flight_quarter'] = df_processed['flight_date'].dt.quarter
            df_processed = df_processed.drop('flight_date', axis=1)
            print(f"  Created: flight_year, month, day, dayofweek, quarter")
        
        # Convert day_of_week_name to numerical
        if 'day_of_week_name' in df_processed.columns:
            day_mapping = {
                'Monday': 1, 'Tuesday': 2, 'Wednesday': 3, 'Thursday': 4,
                'Friday': 5, 'Saturday': 6, 'Sunday': 7
            }
            df_processed['day_of_week_name'] = df_processed['day_of_week_name'].map(day_mapping)
            df_processed['day_of_week_name'] = df_processed['day_of_week_name'].fillna(1)
            print(f"Converted day_of_week_name to numerical")
        
        # Identify column types
        self.categorical_columns = df_processed.select_dtypes(include=['object']).columns.tolist()
        self.numerical_columns = df_processed.select_dtypes(include=['int64', 'float64']).columns.tolist()
        
        print(f"\nCategorical columns ({len(self.categorical_columns)}):")
        for col in self.categorical_columns[:10]:
            unique_vals = df_processed[col].nunique()
            print(f"  {col} ({unique_vals:,} unique values)")
        if len(self.categorical_columns) > 10:
            print(f"  ... and {len(self.categorical_columns) - 10} more")
        
        print(f"\nNumerical columns ({len(self.numerical_columns)}):")
        for col in self.numerical_columns[:15]:
            print(f"  {col}")
        if len(self.numerical_columns) > 15:
            print(f"  ... and {len(self.numerical_columns) - 15} more")
        
        # Encode categorical variables
        print("\nEncoding categorical variables...")
        for col in self.categorical_columns:
            if col in df_processed.columns:
                le = LabelEncoder()
                df_processed[col] = df_processed[col].fillna('Unknown').astype(str)
                df_processed[col] = le.fit_transform(df_processed[col])
                self.label_encoders[col] = le
                print(f"  Encoded: {col}")
        
        # Store feature columns
        self.feature_columns = df_processed.columns.tolist()
        
        print(f"\nTotal features after processing: {len(self.feature_columns):,}")
        print(f"Total samples: {len(df_processed):,}")
        
        return df_processed.values, y_departure, y_arrival
    
    # ==================== ARCHITECTURE 1: ORIGINAL NETWORK ====================
    def build_original_network(self, input_shape):
        """Original architecture from your code"""
        print("\n" + "=" * 70)
        print("BUILDING ORIGINAL NETWORK ARCHITECTURE")
        print("=" * 70)
        print(f"Input shape: ({input_shape},)")
        
        inputs = keras.Input(shape=(input_shape,), name='input_layer')
        
        x = layers.Dense(256, activation='relu', name='dense_1')(inputs)
        x = layers.BatchNormalization(name='batch_norm_1')(x)
        x = layers.Dropout(0.3, name='dropout_1')(x)
        
        x = layers.Dense(128, activation='relu', name='dense_2')(x)
        x = layers.BatchNormalization(name='batch_norm_2')(x)
        x = layers.Dropout(0.3, name='dropout_2')(x)
        
        x = layers.Dense(64, activation='relu', name='dense_3')(x)
        x = layers.BatchNormalization(name='batch_norm_3')(x)
        x = layers.Dropout(0.2, name='dropout_3')(x)
        
        dep_branch = layers.Dense(32, activation='relu', name='dep_dense_1')(x)
        dep_branch = layers.Dense(16, activation='relu', name='dep_dense_2')(dep_branch)
        departure_output = layers.Dense(1, name='departure_delay')(dep_branch)
        
        arr_branch = layers.Dense(32, activation='relu', name='arr_dense_1')(x)
        arr_branch = layers.Dense(16, activation='relu', name='arr_dense_2')(arr_branch)
        arrival_output = layers.Dense(1, name='arrival_delay')(arr_branch)
        
        model = keras.Model(
            inputs=inputs,
            outputs=[departure_output, arrival_output],
            name='original_delay_predictor'
        )
        
        return model
    
    # ==================== ARCHITECTURE 2: WIDE & DEEP ====================
    def build_wide_deep_network(self, input_shape):
        """
        Architecture 1: Wide & Deep Network
        - Wide part: Memorization of feature interactions
        - Deep part: Generalization through hidden layers
        """
        print("\n" + "=" * 70)
        print("BUILDING ARCHITECTURE 1: WIDE & DEEP NETWORK")
        print("=" * 70)
        print(f"Input shape: ({input_shape},)")
        
        inputs = keras.Input(shape=(input_shape,), name='input_layer')
        
        # === WIDE PATH (Memorization) ===
        wide_output = layers.Dense(64, activation='relu', name='wide_dense')(inputs)
        wide_output = layers.Dropout(0.2, name='wide_dropout')(wide_output)
        
        # === DEEP PATH (Generalization) ===
        deep = layers.Dense(512, activation='relu', name='deep_dense_1')(inputs)
        deep = layers.BatchNormalization(name='deep_bn_1')(deep)
        deep = layers.Dropout(0.3, name='deep_dropout_1')(deep)
        
        deep = layers.Dense(256, activation='relu', name='deep_dense_2')(deep)
        deep = layers.BatchNormalization(name='deep_bn_2')(deep)
        deep = layers.Dropout(0.3, name='deep_dropout_2')(deep)
        
        deep = layers.Dense(128, activation='relu', name='deep_dense_3')(deep)
        deep = layers.BatchNormalization(name='deep_bn_3')(deep)
        deep = layers.Dropout(0.2, name='deep_dropout_3')(deep)
        
        # === COMBINE WIDE & DEEP ===
        combined = layers.Concatenate(name='combine_wide_deep')([wide_output, deep])
        
        # Shared representation
        shared = layers.Dense(128, activation='relu', name='shared_dense')(combined)
        shared = layers.Dropout(0.2, name='shared_dropout')(shared)
        
        # Departure branch
        dep_branch = layers.Dense(64, activation='relu', name='dep_dense_1')(shared)
        dep_branch = layers.Dense(32, activation='relu', name='dep_dense_2')(dep_branch)
        dep_branch = layers.Dense(16, activation='relu', name='dep_dense_3')(dep_branch)
        departure_output = layers.Dense(1, name='departure_delay')(dep_branch)
        
        # Arrival branch
        arr_branch = layers.Dense(64, activation='relu', name='arr_dense_1')(shared)
        arr_branch = layers.Dense(32, activation='relu', name='arr_dense_2')(arr_branch)
        arr_branch = layers.Dense(16, activation='relu', name='arr_dense_3')(arr_branch)
        arrival_output = layers.Dense(1, name='arrival_delay')(arr_branch)
        
        model = keras.Model(
            inputs=inputs,
            outputs=[departure_output, arrival_output],
            name='wide_deep_delay_predictor'
        )
        
        return model
    
    # ==================== ARCHITECTURE 3: RESIDUAL NETWORK ====================
    def build_residual_network(self, input_shape):
        """
        Architecture 2: Residual Network (ResNet-style) - Fixed version
        - Skip connections to prevent vanishing gradients
        """
        print("\n" + "=" * 70)
        print("BUILDING ARCHITECTURE 2: RESIDUAL NETWORK")
        print("=" * 70)
        print(f"Input shape: ({input_shape},)")
        
        def residual_block(x, units, dropout_rate=0.2, block_name='res_block'):
            """Residual block with skip connection"""
            shortcut = x
            
            main = layers.Dense(units, activation='relu', name=f'{block_name}_dense1')(x)
            main = layers.BatchNormalization(name=f'{block_name}_bn1')(main)
            main = layers.Dropout(dropout_rate, name=f'{block_name}_dropout1')(main)
            
            main = layers.Dense(units, activation='relu', name=f'{block_name}_dense2')(main)
            main = layers.BatchNormalization(name=f'{block_name}_bn2')(main)
            
            if shortcut.shape[-1] != units:
                shortcut = layers.Dense(units, name=f'{block_name}_shortcut')(shortcut)
            
            output = layers.Add(name=f'{block_name}_add')([main, shortcut])
            output = layers.Activation('relu', name=f'{block_name}_output')(output)
            
            return output
        
        inputs = keras.Input(shape=(input_shape,), name='input_layer')
        
        # Initial projection
        x = layers.Dense(256, activation='relu', name='initial_dense')(inputs)
        x = layers.BatchNormalization(name='initial_bn')(x)
        
        # Residual blocks
        x = residual_block(x, 256, dropout_rate=0.2, block_name='res_block1')
        x = residual_block(x, 256, dropout_rate=0.2, block_name='res_block2')
        
        x = residual_block(x, 128, dropout_rate=0.2, block_name='res_block3')
        x = residual_block(x, 128, dropout_rate=0.2, block_name='res_block4')
        
        x = residual_block(x, 64, dropout_rate=0.1, block_name='res_block5')
        
        # Global pooling - FIXED: Use Reshape instead of Lambda
        x = layers.Reshape((1, 64), name='reshape_for_pooling')(x)
        x = layers.GlobalAveragePooling1D(name='global_pool')(x)
        
        # Departure branch
        dep_branch = layers.Dense(32, activation='relu', name='dep_dense_1')(x)
        dep_branch = layers.Dense(16, activation='relu', name='dep_dense_2')(dep_branch)
        departure_output = layers.Dense(1, name='departure_delay')(dep_branch)
        
        # Arrival branch
        arr_branch = layers.Dense(32, activation='relu', name='arr_dense_1')(x)
        arr_branch = layers.Dense(16, activation='relu', name='arr_dense_2')(arr_branch)
        arrival_output = layers.Dense(1, name='arrival_delay')(arr_branch)
        
        model = keras.Model(
            inputs=inputs,
            outputs=[departure_output, arrival_output],
            name='residual_delay_predictor'
        )
        
        return model
    
    # ==================== ARCHITECTURE 4: TRANSFORMER-STYLE ====================
    def build_transformer_network(self, input_shape):
        """
        Architecture 3: Transformer-style Network
        - Self-attention mechanism for feature interactions
        """
        print("\n" + "=" * 70)
        print("BUILDING ARCHITECTURE 3: TRANSFORMER-STYLE NETWORK")
        print("=" * 70)
        print(f"Input shape: ({input_shape},)")
        
        def transformer_block(x, head_size=32, num_heads=4, ff_dim=128, dropout=0.2, block_name='transformer'):
            """Transformer block with multi-head attention"""
            x_norm = layers.LayerNormalization(epsilon=1e-6, name=f'{block_name}_ln1')(x)
            
            attention_output = layers.MultiHeadAttention(
                num_heads=num_heads,
                key_dim=head_size,
                dropout=dropout,
                name=f'{block_name}_attention'
            )(x_norm, x_norm)
            
            x = layers.Add(name=f'{block_name}_add1')([x, attention_output])
            
            x_norm = layers.LayerNormalization(epsilon=1e-6, name=f'{block_name}_ln2')(x)
            ff_output = layers.Dense(ff_dim, activation='relu', name=f'{block_name}_ff1')(x_norm)
            ff_output = layers.Dropout(dropout, name=f'{block_name}_dropout')(ff_output)
            ff_output = layers.Dense(x.shape[-1], name=f'{block_name}_ff2')(ff_output)
            
            x = layers.Add(name=f'{block_name}_add2')([x, ff_output])
            
            return x
        
        inputs = keras.Input(shape=(input_shape,), name='input_layer')
        
        # Project to higher dimension for attention
        x = layers.Dense(128, activation='relu', name='projection_dense')(inputs)
        x = layers.Reshape((1, 128), name='reshape_for_attention')(x)
        
        # Transformer blocks
        x = transformer_block(x, head_size=32, num_heads=4, ff_dim=256, 
                              dropout=0.2, block_name='transformer1')
        x = transformer_block(x, head_size=32, num_heads=4, ff_dim=256, 
                              dropout=0.2, block_name='transformer2')
        
        # Global pooling
        x = layers.GlobalAveragePooling1D(name='global_pool')(x)
        
        # Shared layers after attention
        x = layers.Dense(128, activation='relu', name='shared_dense_1')(x)
        x = layers.Dropout(0.3, name='shared_dropout_1')(x)
        x = layers.Dense(64, activation='relu', name='shared_dense_2')(x)
        x = layers.Dropout(0.2, name='shared_dropout_2')(x)
        
        # Departure branch
        dep_branch = layers.Dense(32, activation='relu', name='dep_dense_1')(x)
        dep_branch = layers.Dense(16, activation='relu', name='dep_dense_2')(dep_branch)
        departure_output = layers.Dense(1, name='departure_delay')(dep_branch)
        
        # Arrival branch
        arr_branch = layers.Dense(32, activation='relu', name='arr_dense_1')(x)
        arr_branch = layers.Dense(16, activation='relu', name='arr_dense_2')(arr_branch)
        arrival_output = layers.Dense(1, name='arrival_delay')(arr_branch)
        
        model = keras.Model(
            inputs=inputs,
            outputs=[departure_output, arrival_output],
            name='transformer_delay_predictor'
        )
        
        return model
    
    def build_neural_network(self, input_shape):
        """Build the selected neural network architecture."""
        if self.model_type == 'original':
            return self.build_original_network(input_shape)
        elif self.model_type == 'wide_deep':
            return self.build_wide_deep_network(input_shape)
        elif self.model_type == 'residual':
            return self.build_residual_network(input_shape)
        elif self.model_type == 'transformer':
            return self.build_transformer_network(input_shape)
        else:
            print(f"Unknown model type '{self.model_type}'. Using Wide & Deep.")
            return self.build_wide_deep_network(input_shape)
    
    def train_and_evaluate(self, df=None, epochs=15, batch_size=2048, 
                          validation_split=0.2, comparison_tracker=None, 
                          model_name=None):
        """
        Train model and collect comprehensive metrics for comparison.
        """
        import time
        
        if model_name is None:
            model_name = self.model_type.replace('_', ' ').title()
        
        print("\n" + "█" * 70)
        print(f"█ TRAINING MODEL: {model_name}")
        print("█" * 70)
        
        # Load data if not provided
        if df is None:
            df = self.load_and_prepare_data()
        
        # Prepare features
        X, y_dep, y_arr = self.prepare_features(df)
        
        # Split data
        X_train, X_temp, y_dep_train, y_dep_temp, y_arr_train, y_arr_temp = train_test_split(
            X, y_dep, y_arr, test_size=0.3, random_state=42
        )
        
        X_val, X_test, y_dep_val, y_dep_test, y_arr_val, y_arr_test = train_test_split(
            X_temp, y_dep_temp, y_arr_temp, test_size=0.5, random_state=42
        )
        
        # Scale features
        print("\nScaling features...")
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_val_scaled = self.scaler.transform(X_val)
        X_test_scaled = self.scaler.transform(X_test)
        
        print(f"\nDataset sizes:")
        print(f"  Training:   {X_train_scaled.shape[0]:,} samples")
        print(f"  Validation: {X_val_scaled.shape[0]:,} samples")
        print(f"  Test:       {X_test_scaled.shape[0]:,} samples")
        print(f"  Features:   {X_train_scaled.shape[1]:,}")
        
        # Build model
        self.model = self.build_neural_network(X_train_scaled.shape[1])
        
        # Optimizer with float learning rate
        optimizer = keras.optimizers.Adam(learning_rate=0.001)
        
        # Compile model
        self.model.compile(
            optimizer=optimizer,
            loss={
                'departure_delay': 'mse',
                'arrival_delay': 'mse'
            },
            loss_weights={
                'departure_delay': 1.0,
                'arrival_delay': 1.0
            },
            metrics={
                'departure_delay': ['mae', keras.metrics.RootMeanSquaredError()],
                'arrival_delay': ['mae', keras.metrics.RootMeanSquaredError()]
            }
        )
        
        # Display model architecture
        print("\nModel Summary:")
        self.model.summary()
        
        # Callbacks
        callbacks = [
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=3,
                restore_best_weights=True,
                min_delta=0.001,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=2,
                min_lr=0.00001,
                verbose=1,
                cooldown=1
            ),
            keras.callbacks.ModelCheckpoint(
                f'best_model_{self.model_type}.h5',
                monitor='val_loss',
                save_best_only=True,
                verbose=0
            )
        ]
        
        # Train and measure training time
        print("\nStarting training...")
        start_time = time.time()
        
        self.history = self.model.fit(
            X_train_scaled,
            {'departure_delay': y_dep_train, 'arrival_delay': y_arr_train},
            validation_data=(X_val_scaled, 
                           {'departure_delay': y_dep_val, 'arrival_delay': y_arr_val}),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )
        
        training_time = time.time() - start_time
        
        # If comparison tracker is provided, evaluate and store results
        if comparison_tracker is not None:
            comparison_tracker.evaluate_model(
                model=self.model,
                X_test=X_test_scaled,
                y_dep_test=y_dep_test,
                y_arr_test=y_arr_test,
                model_name=model_name,
                training_time=training_time
            )
        
        return self.history, training_time, (X_test_scaled, y_dep_test, y_arr_test)
    
    def get_unique_values_from_dataset(self, df, column_name):
        """Get unique values from a specific column."""
        if column_name in df.columns:
            unique_vals = df[column_name].dropna().unique()
            return sorted(unique_vals.tolist())
        return []
    
    def prepare_user_input_from_dataset(self, df, user_params):
        """Prepare user input using actual values from the dataset."""
        user_df = pd.DataFrame(columns=self.feature_columns)
        user_df.loc[0] = 0
        
        param_mapping = {
            'airline': 'airline_name',
            'departure': 'dep_airport_code',
            'arrival': 'arr_airport_code',
            'weekday': 'day_of_week_num',
            'manufacturer': 'aircraft_manufacturer',
            'model': 'aircraft_model',
            'flight_date': 'flight_date',
            'dep_time_label': 'dep_time_label',
            'flight_duration': 'flight_duration_min',
            'distance_type': 'flight_distance_type',
            'aircraft_age': 'aircraft_age_years',
            'tail_number': 'aircraft_tail_num',
            'dep_temp_avg': 'dep_temp_avg',
            'dep_temp_min': 'dep_temp_min',
            'dep_temp_max': 'dep_temp_max',
            'dep_precip': 'dep_precip',
            'dep_snow': 'dep_snow',
            'dep_wind_dir': 'dep_wind_dir',
            'dep_wind_speed': 'dep_wind_speed',
            'dep_pressure': 'dep_pressure',
            'arr_temp_avg': 'arr_temp_avg',
            'arr_temp_min': 'arr_temp_min',
            'arr_temp_max': 'arr_temp_max',
            'arr_precip': 'arr_precip',
            'arr_snow': 'arr_snow',
            'arr_wind_dir': 'arr_wind_dir',
            'arr_wind_speed': 'arr_wind_speed',
            'arr_pressure': 'arr_pressure',
            'delay_carrier': 'delay_carrier_min',
            'delay_weather': 'delay_weather_min',
            'delay_nas': 'delay_nas_min',
            'delay_security': 'delay_security_min',
            'delay_last_aircraft': 'delay_last_aircraft_min',
            'route_distance': 'route_distance_km',
            'dep_city': 'dep_city_name',
            'arr_city': 'arr_city_name',
            'dep_state': 'dep_state',
            'arr_state': 'arr_state',
            'dep_airport_name': 'dep_airport_name',
            'arr_airport_name': 'arr_airport_name'
        }
        
        for user_key, dataset_col in param_mapping.items():
            if user_key in user_params and dataset_col in user_df.columns:
                user_df.loc[0, dataset_col] = user_params[user_key]
            elif dataset_col in df.columns:
                if df[dataset_col].dtype in ['int64', 'float64']:
                    user_df.loc[0, dataset_col] = df[dataset_col].median()
                else:
                    most_common = df[dataset_col].mode()
                    if not most_common.empty:
                        user_df.loc[0, dataset_col] = most_common.iloc[0]
        
        if 'flight_date' in user_params:
            try:
                flight_date = pd.to_datetime(user_params['flight_date'])
                user_df.loc[0, 'flight_year'] = flight_date.year
                user_df.loc[0, 'flight_month'] = flight_date.month
                user_df.loc[0, 'flight_day'] = flight_date.day
                user_df.loc[0, 'flight_dayofweek'] = flight_date.dayofweek
                user_df.loc[0, 'flight_quarter'] = flight_date.quarter
            except:
                if 'flight_date' in df.columns:
                    median_date = pd.to_datetime(df['flight_date']).median()
                    user_df.loc[0, 'flight_year'] = median_date.year
                    user_df.loc[0, 'flight_month'] = median_date.month
                    user_df.loc[0, 'flight_day'] = median_date.day
                    user_df.loc[0, 'flight_dayofweek'] = median_date.dayofweek
                    user_df.loc[0, 'flight_quarter'] = median_date.quarter
        
        if 'weekday' in user_params and 'day_of_week_name' in user_df.columns:
            user_df.loc[0, 'day_of_week_name'] = user_params['weekday']
        
        return user_df
    
    def predict(self, user_params):
        """Make delay predictions for user-provided flight details."""
        if self.model is None:
            raise ValueError("Model must be trained before making predictions. Call train() first.")
        
        print("=" * 70)
        print("MAKING PREDICTIONS")
        print("=" * 70)
        
        df = self.load_and_prepare_data()
        
        print("\nAvailable options from dataset:")
        airlines = self.get_unique_values_from_dataset(df, 'airline_name')[:10]
        airports = self.get_unique_values_from_dataset(df, 'dep_airport_code')[:10]
        manufacturers = self.get_unique_values_from_dataset(df, 'aircraft_manufacturer')[:5]
        models = self.get_unique_values_from_dataset(df, 'aircraft_model')[:10]
        
        print(f"Airlines: {airlines}")
        print(f"Airports: {airports}")
        print(f"Manufacturers: {manufacturers}")
        print(f"Models: {models}")
        
        user_df = self.prepare_user_input_from_dataset(df, user_params)
        
        print("\n" + "-" * 50)
        print("USER INPUT SUMMARY")
        print("-" * 50)
        
        required_params = ['airline', 'departure', 'arrival', 'weekday', 
                          'manufacturer', 'model']
        
        for param in required_params:
            if param in user_params:
                print(f"{param.capitalize():15}: {user_params[param]}")
            else:
                print(f"{param.capitalize():15}: Using dataset default")
        
        print("\nWeather Conditions (using dataset patterns):")
        weather_params = {
            'Departure': ['dep_temp_avg', 'dep_precip', 'dep_wind_speed'],
            'Arrival': ['arr_temp_avg', 'arr_precip', 'arr_wind_speed']
        }
        
        for location, params in weather_params.items():
            print(f"\n{location}:")
            for param in params:
                if param in user_df.columns:
                    value = user_df.loc[0, param]
                    if 'temp' in param:
                        print(f"  {param}: {value:.1f}°F")
                    elif 'precip' in param:
                        print(f"  {param}: {value:.2f} in")
                    elif 'wind' in param:
                        print(f"  {param}: {value:.1f} mph")
        
        df_processed = user_df.copy()
        
        for col in self.categorical_columns:
            if col in df_processed.columns and col in self.label_encoders:
                le = self.label_encoders[col]
                value = str(df_processed.loc[0, col])
                if value in le.classes_:
                    df_processed.loc[0, col] = le.transform([value])[0]
                else:
                    df_processed.loc[0, col] = 0
        
        df_processed = df_processed[self.feature_columns]
        X_scaled = self.scaler.transform(df_processed.values)
        
        print("\n" + "-" * 50)
        print("NEURAL NETWORK PREDICTION")
        print("-" * 50)
        
        predictions = self.model.predict(X_scaled, verbose=0)
        
        departure_delay = predictions[0].flatten()[0]
        arrival_delay = predictions[1].flatten()[0]
        
        print("\n" + "=" * 70)
        print("PREDICTION RESULTS")
        print("=" * 70)
        
        print(f"\nPredicted Departure Delay: {departure_delay:.1f} minutes")
        print(f"Predicted Arrival Delay: {arrival_delay:.1f} minutes")
        print(f"Additional Delay In-flight: {arrival_delay - departure_delay:.1f} minutes")
        
        print("\nInterpretation:")
        
        def interpret_delay(delay, delay_type):
            if delay < -15:
                return f"✓ {delay_type}: Very Early ({abs(delay):.0f} minutes early)"
            elif delay < 0:
                return f"✓ {delay_type}: Early ({abs(delay):.0f} minutes early)"
            elif delay <= 15:
                return f"✓ {delay_type}: On Time ({delay:.0f} minutes delay)"
            elif delay <= 60:
                return f"⚠ {delay_type}: Moderately Delayed ({delay:.0f} minutes)"
            elif delay <= 120:
                return f"⚠ {delay_type}: Significantly Delayed ({delay:.0f} minutes)"
            else:
                return f"✗ {delay_type}: Severely Delayed ({delay:.0f} minutes)"
        
        print(interpret_delay(departure_delay, "Departure"))
        print(interpret_delay(arrival_delay, "Arrival"))
        
        print("\n" + "-" * 50)
        print("KEY FACTORS INFLUENCING PREDICTION")
        print("-" * 50)
        
        if 'dep_precip' in user_df.columns and user_df.loc[0, 'dep_precip'] > 0.5:
            print("• Heavy precipitation at departure airport")
        if 'dep_wind_speed' in user_df.columns and user_df.loc[0, 'dep_wind_speed'] > 20:
            print("• Strong winds at departure airport")
        if 'arr_precip' in user_df.columns and user_df.loc[0, 'arr_precip'] > 0.3:
            print("• Precipitation at arrival airport")
        
        delay_factors = [
            ('delay_carrier_min', 'Carrier-related issues', 20),
            ('delay_weather_min', 'Weather conditions', 15),
            ('delay_nas_min', 'National Air System delays', 20),
        ]
        
        for col, description, threshold in delay_factors:
            if col in user_df.columns and user_df.loc[0, col] > threshold:
                print(f"• {description}: {user_df.loc[0, col]:.0f} minutes")
        
        if 'dep_time_label' in user_params:
            time_label = user_params['dep_time_label']
            if time_label in ['Evening', 'Night']:
                print("• Evening/Night departure (typically higher congestion)")
            elif time_label == 'Morning':
                print("• Morning departure (typically better on-time performance)")
        
        return departure_delay, arrival_delay
    
    def save_model(self, filepath='flight_delay_model'):
        """Save the trained model and preprocessing objects."""
        import joblib
        import os
        
        filepath = f'{filepath}_{self.model_type}'
        os.makedirs(filepath, exist_ok=True)
        
        self.model.save(f'{filepath}/model.h5')
        joblib.dump(self.scaler, f'{filepath}/scaler.pkl')
        joblib.dump(self.label_encoders, f'{filepath}/label_encoders.pkl')
        joblib.dump(self.feature_columns, f'{filepath}/feature_columns.pkl')
        joblib.dump(self.categorical_columns, f'{filepath}/categorical_columns.pkl')
        
        print(f"\nModel saved to '{filepath}/'")
    
    def load_model(self, filepath='flight_delay_model'):
        """Load a previously trained model."""
        import joblib
        from tensorflow.keras.models import load_model
        
        filepath = f'{filepath}_{self.model_type}'
        
        self.model = load_model(f'{filepath}/model.h5')
        self.scaler = joblib.load(f'{filepath}/scaler.pkl')
        self.label_encoders = joblib.load(f'{filepath}/label_encoders.pkl')
        self.feature_columns = joblib.load(f'{filepath}/feature_columns.pkl')
        self.categorical_columns = joblib.load(f'{filepath}/categorical_columns.pkl')
        
        print(f"\nModel loaded from '{filepath}/'")


# ============================================================================
# MAIN EXECUTION - RUN COMPREHENSIVE COMPARISON
# ============================================================================

def run_full_comparison():
    """
    Run comprehensive comparison of all model architectures.
    """
    print("\n" + "=" * 80)
    print("🚀 FLIGHT DELAY PREDICTION - COMPREHENSIVE MODEL COMPARISON")
    print("=" * 80)
    print("\n📊 Dataset: 6.7M flight records")
    print("📈 Models: Original NN, Wide & Deep, Residual Network, Transformer")
    print("⏱️  Epochs: 15 | Batch Size: 2048")
    print("📊 Metrics: MAE, RMSE, R², MAPE, SMAPE, Training Time, Prediction Speed")
    print("=" * 80)
    
    # Initialize comparison tracker
    comparator = ModelComparison()
    
    # Model configurations to test
    model_configs = [
        {'type': 'original', 'name': 'Original Neural Network'},
        {'type': 'wide_deep', 'name': 'Wide & Deep Network'},
        {'type': 'residual', 'name': 'Residual Network'},
        {'type': 'transformer', 'name': 'Transformer Network'}
    ]
    
    # Load data once (shared across all models)
    print("\n📂 Loading dataset...")
    temp_predictor = FlightDelayNeuralNetwork()
    df = temp_predictor.load_and_prepare_data()
    
    # Train and evaluate each model
    for config in model_configs:
        predictor = FlightDelayNeuralNetwork(model_type=config['type'])
        
        # Train with comparison tracking
        history, training_time, test_data = predictor.train_and_evaluate(
            df=df,
            epochs=15,
            batch_size=2048,
            comparison_tracker=comparator,
            model_name=config['name']
        )
    
    # Display comprehensive comparison table
    print("\n" + "=" * 80)
    print("📊 GENERATING COMPREHENSIVE COMPARISON REPORT (WITH SMAPE)")
    print("=" * 80)
    
    comparator.display_comparison_table()
    
    # Generate plots
    print("\n📈 Generating visualization plots with MAPE vs SMAPE...")
    comparator.plot_comparison()
    
    # Save reports
    comparator.save_comparison_report('flight_delay_model_comparison')
    
    print("\n" + "=" * 80)
    print("✅ COMPARISON COMPLETE!")
    print("=" * 80)
    print("\n📁 Output files generated:")
    print("   • flight_delay_model_comparison.csv - Raw metrics data (with SMAPE)")
    print("   • flight_delay_model_comparison_summary.txt - Detailed report (with SMAPE)")
    print("   • best_model_*.h5 - Saved model files for each architecture")
    print("\n🏆 Check the summary table above for the best performing model!")
    
    return comparator


def run_single_model():
    """
    Run a single model for testing or deployment.
    """
    print("\n" + "=" * 80)
    print("✈️  FLIGHT DELAY PREDICTION - SINGLE MODEL TRAINING")
    print("=" * 80)
    
    # Choose model type
    model_type = 'wide_deep'  # Change this to try different architectures
    
    print(f"\n📋 Selected Model: {model_type.replace('_', ' ').title()}")
    
    # Initialize predictor
    predictor = FlightDelayNeuralNetwork(model_type=model_type)
    
    # Train model
    history, training_time, test_data = predictor.train_and_evaluate(
        epochs=15,
        batch_size=2048,
        model_name=f"{model_type.replace('_', ' ').title()}"
    )
    
    # Save model
    predictor.save_model('trained_model')
    
    # Example prediction
    print("\n" + "=" * 80)
    print("🎯 EXAMPLE PREDICTION")
    print("=" * 80)
    
    user_input = {
        'airline': 'Delta Air Lines',
        'departure': 'ATL',
        'arrival': 'LAX',
        'weekday': 3,
        'manufacturer': 'Boeing',
        'model': '737-800',
        'flight_date': '2024-01-15',
        'dep_time_label': 'Morning',
        'flight_duration': 245,
        'distance_type': 'Long',
        'aircraft_age': 8,
        'dep_temp_avg': 68,
        'dep_precip': 0.1,
        'dep_wind_speed': 12,
        'arr_temp_avg': 72,
        'arr_precip': 0.0,
        'arr_wind_speed': 8,
    }
    
    try:
        dep_delay, arr_delay = predictor.predict(user_input)
    except Exception as e:
        print(f"\n⚠️  Prediction example requires trained model. Error: {e}")
    
    return predictor


if __name__ == "__main__":
    # Check TensorFlow setup
    print("TensorFlow version:", tf.__version__)
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        print(f"GPU Available: {len(gpus)} device(s)")
        for gpu in gpus:
            print(f"  - {gpu.name}")
        # Enable memory growth
        for gpu in gpus:
            try:
                tf.config.experimental.set_memory_growth(gpu, True)
                print(f"  Memory growth enabled for {gpu.name}")
            except:
                pass
    else:
        print("⚠️  GPU Available: No, using CPU")
        print("   Warning: Training 6.7M records on CPU will be very slow!")
    
    print("\n" + "═" * 80)
    print("⚡ FLIGHT DELAY PREDICTION SYSTEM - 6.7M RECORDS")
    print("═" * 80)
    
    # Menu for execution mode
    print("\nSelect execution mode:")
    print("  1. Full Comparison (Train all 4 models with SMAPE)")
    print("  2. Single Model (Train one model)")
    print("  3. Exit")
    
    choice = input("\nEnter choice (1-3): ").strip()
    
    if choice == '1':
        # Run full comparison of all models
        comparator = run_full_comparison()
    elif choice == '2':
        # Run single model
        predictor = run_single_model()
    else:
        print("\n👋 Exiting. Goodbye!")
```
